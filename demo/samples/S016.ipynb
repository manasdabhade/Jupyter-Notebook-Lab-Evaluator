# Lab 3: Text classification

## Data
Twenty-four short movie reviews labelled positive or negative.

```python
reviews = [("I loved this great movie", 1), ("Wonderful acting and a great story", 1), ("A brilliant and wonderful film", 1),
    ("Great fun, loved the cast", 1), ("Brilliant music and wonderful visuals", 1), ("Loved it, a great and brilliant script", 1),
    ("Wonderful, fun and great performances", 1), ("A great film, loved every scene", 1), ("Brilliant story and great direction", 1),
    ("Fun, wonderful and loved by everyone", 1), ("Great acting, brilliant ending", 1), ("Loved the wonderful characters", 1),
    ("I hated this boring movie", 0), ("Terrible acting and a boring story", 0), ("An awful and dull film", 0),
    ("Boring mess, hated the cast", 0), ("Awful music and terrible visuals", 0), ("Hated it, a dull and awful script", 0),
    ("Terrible, dull and boring performances", 0), ("A boring film, hated every scene", 0), ("Awful story and terrible direction", 0),
    ("Dull, terrible and hated by everyone", 0), ("Boring acting, awful ending", 0), ("Hated the dull characters", 0)]
texts, labels = [t for t, _ in reviews], [y for _, y in reviews]
```

## Pre-processing
Tokenize, remove stop words and stem.

```python
import re
from nltk.stem import PorterStemmer
STOP = {"i", "the", "a", "and", "this", "it", "by", "every"}
ps = PorterStemmer()

def clean_text(text):
    """Lowercase, tokenize, remove stop words and stem each word."""
    words = re.findall(r"[a-z]+", text.lower())      # tokenize
    return [ps.stem(w) for w in words if w not in STOP]  # drop stop words, stem the rest

print(clean_text("Running runs!"))
```

## Model
TF-IDF features and a Naive Bayes classifier.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# hold out 25% of reviews for testing
X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.25, random_state=0, stratify=labels)
vectorizer = TfidfVectorizer(analyzer=clean_text)   # TF-IDF features from cleaned words
model = MultinomialNB().fit(vectorizer.fit_transform(X_train), y_train)
print("Accuracy:", accuracy_score(y_test, model.predict(vectorizer.transform(X_test))))
```

## Explanation
Each review is tokenized, stop words are removed and every word is reduced to its stem with the Porter stemmer, so running and runs become the same word. TF-IDF then turns each review into a vector where a word's frequency is multiplied by its inverse document frequency, so words found in many reviews count for less. A Multinomial Naive Bayes classifier uses Bayes' theorem, assuming words are independent given the class, and picks the sentiment with the highest probability. Accuracy is measured on the held-out test reviews.
