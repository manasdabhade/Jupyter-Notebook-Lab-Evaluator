# Lab: News headline topic classification

## Load the data

```python
data = [('striker scores twice as united win the derby', 'sport'), ('coach praises defence after cup victory', 'sport'), ('tennis champion reaches the final in straight sets', 'sport'), ('cricket team announces squad for test series', 'sport'), ('goalkeeper saves penalty in late drama', 'sport'), ('olympic sprinter breaks national record', 'sport'), ('league title race goes to the final match', 'sport'), ('captain injured ahead of world cup qualifier', 'sport'), ('central bank raises interest rates again', 'business'), ('shares fall as quarterly profits miss forecasts', 'business'), ('startup raises funding to expand into europe', 'business'), ('inflation slows for the third month', 'business'), ('retailer reports record holiday sales', 'business'), ('oil prices climb on supply concerns', 'business'), ('bank announces merger with regional lender', 'business'), ('stock market closes at record high', 'business'), ('new smartphone launches with faster chip', 'tech'), ('researchers unveil open source language model', 'tech'), ('software update fixes security flaw in browser', 'tech'), ('chipmaker reveals next generation processor', 'tech'), ('app store policy changes anger developers', 'tech'), ('cloud outage disrupts popular websites', 'tech'), ('robotics company demos warehouse robot', 'tech'), ('new laptop promises longer battery life', 'tech')]  # (headline, topic)
headlines = [h for h, _ in data]
topics = [t for _, t in data]
print(len(headlines), 'headlines')
```

## Step 1: TF-IDF features

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(stop_words='english')  # weight words by rarity
X = vectorizer.fit_transform(headlines)
print(X.shape)
```

## Step 2: Train a linear SVM

```python
from sklearn.svm import LinearSVC
from sklearn.model_selection import cross_val_score
model = LinearSVC()  # maximum-margin linear classifier
scores = cross_val_score(model, X, topic, cv=4)  # 4-fold cross-validation
print('Mean accuracy:', round(scores.mean(), 2))
model.fit(X, topics)
```

## Step 3: Classify new headlines

```python
new = ['striker scores in cup final', 'bank cuts rates', 'chip shortage hits phones']  # unseen
for h, p in zip(new, model.predict(vectorizer.transform(new))):
    print(h, '->', p)
```
