# Lab 5: K-Means

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.8, random_state=42)  # 3 groups of points
km = KMeans(n_clusters=3, n_init=10, random_state=42)  # 3 centroids, 10 restarts
labels = km.fit_predict(X)
print("Silhouette:", round(silhouette_score(X, labels), 3))
```

## Explanation
We use agglomerative hierarchical clustering: each point starts as its own cluster and the two closest clusters are merged again and again, building a dendrogram that we cut into three clusters. The silhouette score shows the clusters are well separated.
