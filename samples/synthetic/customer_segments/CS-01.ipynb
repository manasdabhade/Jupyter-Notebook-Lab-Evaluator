# Lab: Customer segmentation

## Load the data

```python
from sklearn.datasets import make_blobs
import numpy as np
X, _ = make_blobs(n_samples=150, centers=[[20, 30], [45, 70], [70, 25]], cluster_std=5, random_state=7)  # (age, spend score)
print(X.shape)
```

## Step 1: Scale the features

```python
from sklearn.preprocessing import StandardScaler
X_scaled = StandardScaler().fit_transform(X)  # mean 0, variance 1
print(X_scaled.mean(axis=0).round(2))
```

## Step 2: Agglomerative clustering

```python
from sklearn.cluster import AgglomerativeClustering
model = AgglomerativeClustering(n_clusters=3, linkage='ward')  # merge clusters bottom-up
labels = model.fit_predict(X_scaled)
print(len(set(labels)), 'clusters')
```

## Step 3: Evaluate

```python
from sklearn.metrics import silhouette_score
print('Silhouette:', round(silhouette_score(X_scaled, labels), 2))  # -1 to 1
```

## Step 4: Describe the segments

```python
def segment_means(X, labels):
    """Average age and spend score of each cluster."""
    return {int(k): X[labels == k].mean(axis=0).round(1).tolist() for k in sorted(set(labels))}
print(segment_means(X, labels))
```

## Explanation
The two features, age and spending score, are standardised so both have mean 0 and variance 1 and neither dominates the distance. We then use agglomerative hierarchical clustering: every customer starts as its own cluster and, at each step, the two closest clusters are merged. Ward linkage chooses the merge that least increases the total within-cluster variance. Repeating this builds a tree of merges, a dendrogram, and cutting the tree at three clusters gives the segments. The silhouette score compares how close each point is to its own cluster versus the nearest other cluster; values near 1 mean well separated segments. The segment averages describe a young low spending group, a middle aged high spending group and an older low spending group.
