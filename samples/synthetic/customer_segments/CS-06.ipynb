# Lab: Customer segmentation

## Load the data

```python
from sklearn.datasets import make_blobs
import numpy as np
X, _ = make_blobs(n_samples=150, centers=[[20, 30], [45, 70], [70, 25]], cluster_std=5, random_state=7)  # (age, spend score)
print(X.shape)
```

## Step 1: Scale the features

```python
from sklearn.preprocessing import StandardScaler
X_scaled = StandardScaler().fit_transform(X)  # mean 0, variance 1
print(X_scaled.mean(axis=0).round(2))
```

## Step 2: Agglomerative clustering

```python
from sklearn.cluster import AgglomerativeClustering
model = AgglomerativeClustering(n_clusters=3, linkage='ward')  # merge clusters bottom-up
labels = model.fit_predict(X_scaled)
print(len(set(labels)), 'clusters')
```

## Step 3: Evaluate

```python
from sklearn.metrics import silhouette_score
print('Silhouette:', round(silhouette_score(X_scaled, labels), 2))  # -1 to 1
```

## Step 4: Describe the segments

```python
def segment_means(X, labels):
    """Average age and spend score of each cluster."""
    return {int(k): X[labels == k].mean(axis=0).round(1).tolist() for k in sorted(set(labels))}
print(segment_means(X, labels))
```

## Explanation
Age and spending are put on the same scale so one does not outweigh the other. Each customer begins in a group of one, and the two nearest groups are joined again and again, always choosing the join that keeps groups tight. Stopping when three groups remain gives the segments. A score between minus one and one tells us how much closer customers are to their own group than to the next group, and higher is better. The group averages describe who is in each segment.
